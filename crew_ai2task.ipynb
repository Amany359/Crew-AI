```python
from crewai import Crew, Process, Agent, Task, LLM


llm = LLM(
    model="gemini/gemini-2.0-flash",
    temperature=0,
)

# Writer Agent
writer_agent = Agent(
    role="كاتب مقالات",
    goal="كتابة مقالات تقنية طويلة ومُنظمة باللغة العربية.",
    backstory="كاتب تقني محترف لديه خبرة في تبسيط المفاهيم البرمجية للقُرّاء العرب.",
    llm=llm,
    verbose=True
)

writer_task = Task(
    description="اكتب مقالًا طويلًا ومُنظمًا عن {topic} باللغة العربية، يتضمن مقدمة، شرح المزايا الرئيسية، وأهم حالات الاستخدام.",
    expected_output="مقال لا يقل عن 100 كلمة، منظم بعناوين وفقرات باللغة العربية.",
    agent=writer_agent
)

# Summarizer Agent
summarizer_agent = Agent(
    role="ملخص مقالات",
    goal="تلخيص المقالات بشكل موجز وواضح باللغة العربية.",
    backstory="خبير في تلخيص المحتوى وتحويله إلى نقاط رئيسية مفهومة.",
    llm=llm,
    verbose=True
)

summarizer_task = Task(
    description="لخص المقال المكتوب إلى نقاط رئيسية + اكتب ملخص قصير (TL;DR) بالعربية.",
    expected_output="ملخص في 5 نقاط كحد أقصى، بالإضافة إلى TL;DR لا يتجاوز 20 كلمة.",
    agent=summarizer_agent
)

# Fact-Checker Agent
fact_checker_agent = Agent(
    role="مراجع تقني",
    goal="مراجعة المحتوى للتأكد من دقته التقنية باللغة العربية.",
    backstory="مهندس برمجيات بخبرة عميقة في أطر عمل بايثون مثل Django.",
    llm=llm,
    verbose=True
)

fact_checker_task = Task(
    description="راجع المقال للتأكد من صحته التقنية، وعدّل أو صحح أي معلومات خاطئة.",
    expected_output="مقال مُراجع تقنيًا، أو تقرير يؤكد أن المعلومات صحيحة.",
    agent=fact_checker_agent
)

# Metadata Agent
metadata_agent = Agent(
    role="خبير SEO",
    goal="إنشاء عناوين ووصف وكلمات مفتاحية صديقة لمحركات البحث باللغة العربية.",
    backstory="خبير تحسين محركات البحث (SEO) يركز على المحتوى العربي.",
    llm=llm,
    verbose=True
)

metadata_task = Task(
    description="أنشئ بيانات SEO للمقال: عنوان أقل من 60 حرفًا، وصف أقل من 160 حرفًا، و5-10 كلمات مفتاحية بالعربية.",
    expected_output="عنوان + وصف + قائمة كلمات مفتاحية.",
    agent=metadata_agent
)

# Crew
crew = Crew(
    agents=[
        writer_agent,
        summarizer_agent,
        fact_checker_agent,
        metadata_agent
    ],
    tasks=[
        writer_task,
        summarizer_task,
        fact_checker_task,
        metadata_task
    ],
    process=Process.sequential
)

result = crew.kickoff(
    inputs={
        "topic": "إطار عمل Django"
    }
)

print(result)
```
